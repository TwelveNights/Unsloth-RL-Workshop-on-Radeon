# Part 0: Foundational Concepts

## Problem statement

A pre-trained language model such as Gemma 4 is a generalist: it has processed a large volume of text and can converse, summarize, and write code. It has not, however, practiced any specific task against direct feedback on its own output.

The task addressed in this notebook is deliberately concrete and easy to verify:

> Write a Python function `strategy(board, initial)` that examines a Sudoku board and returns the next move as `(row, col, number)`.

This task can be verified mechanically: no human judgment is required to determine whether an answer is good. The function the model writes can be run on a real puzzle, and the number of legal moves it makes can be counted directly. This property, verifiability, is what makes the task suitable for reinforcement learning.

### Sudoku rules

A Sudoku puzzle consists of filling a 9x9 grid so that every row, every column, and every 3x3 box contains the digits 1 through 9 exactly once. The puzzle begins with some cells already filled; these given cells are fixed and cannot be altered.

In the completed grid shown below, the black digits are the given clues the puzzle started with, and the red digits are the moves required to complete it. `initial` corresponds to the black digits, which a strategy must never overwrite; `board` is the grid in its current state. Each red digit corresponds to one call to `place_number(row, col, num)`, which is legal only if it violates none of the three rules stated above.

## Supervised fine-tuning versus reinforcement learning

There are two general approaches to teaching a model a new task.

**Supervised fine-tuning (SFT)** is a form of imitation learning: a set of examples of correct answers is collected, and the model learns to reproduce them. It is the standard starting point for fine-tuning. Its cost is that a correct answer must be produced for every training example.

**Reinforcement learning (RL)** is a form of learning by practice: instead of supplying the answer, a way to score an answer is supplied. The model generates attempts, each attempt is scored, and the model's parameters are updated to make well-scoring attempts more likely.

RL is the natural fit for the Sudoku task considered here. Writing hundreds of correct, diverse Sudoku solvers by hand would be impractical, whereas writing a function that checks whether a given solver works is straightforward.

<img alt="Supervised fine-tuning copies a human-written gold answer, while GRPO lets the model write several attempts and rewards the ones that score best" src="" width="100%">

*We never show the model a correct solver. It writes code, we run the code, and we reward what works.*

## What is GRPO?

**GRPO (Group Relative Policy Optimization)** is defined as a reinforcement learning algorithm that estimates the advantage of a candidate answer by comparing it to other candidates generated for the same prompt, rather than by using a separate value network. It is a memory-efficient method, popularized by the DeepSeek reasoning work, and is particularly effective at training models to reason.

The name describes the mechanism:

- **Group.** For a single prompt, the model generates a group of several candidate answers rather than one.
- **Relative.** Each candidate is scored and then judged relative to the other candidates in its own group. A candidate that scores above the group average is reinforced; one that scores below the average is discouraged.
- **Policy optimization.** The "policy" refers to the model's behavior. The policy is optimized by shifting it toward the better-scoring candidates.

The comparison being relative, rather than absolute, is the central idea. Earlier RL methods require a second neural network (a value model, or critic) to estimate how good an answer should be, at additional cost in memory and complexity. GRPO instead uses the group average as the baseline, eliminating the need for a critic network. This reduction in memory requirements is what allows the present exercise to run on a single GPU.

<img alt="The four stages of one GRPO step: generate a group of candidates, score each with the reward functions, compare scores within the group, then update the LoRA adapters toward the above-average candidates" src="" width="100%">

*One pass of this loop is one training step.*

## Reward functions

In supervised fine-tuning, effort is concentrated on constructing the dataset. In reinforcement learning, effort is concentrated on designing the reward function, since it is the sole determinant of what "better" means; it is accordingly the most important design decision in this notebook.

Three reward functions are defined below, and their scores are summed:

1. **`function_works`** rewards output that is valid, runnable Python code defining a `strategy` function. Without this term, the model has no incentive to produce parseable code at all.
2. **`no_cheating`** penalizes the model for importing external libraries; the puzzle must be solved using plain Python.
3. **`strategy_succeeds`** is the primary objective. The generated function is executed against an actual Sudoku puzzle and scored according to the number of legal moves it makes, with a substantial bonus for solving the puzzle outright.

Rewards 1 and 2 function as guardrails; reward 3 defines the goal. The total score for an attempt is the sum of all three.

<img alt="Three reward functions: function_works checks the code compiles, no_cheating penalizes imports, and strategy_succeeds runs the strategy on a real puzzle and scores the legal moves it makes" src="" width="100%">

*Guardrails first, then the goal.*

## Reward hacking and the need for a sandbox

A model optimizes the reward function as written, not the goal that was intended. If a cheap shortcut to a high score exists, the model will find it. This phenomenon is known as **reward hacking**. It does not represent a malfunction of the model; rather, the model is doing precisely what it was asked to do, which reveals that the reward function specified the wrong objective.

In the present task, the obvious shortcut is to `import` an existing Sudoku solver, or otherwise access code elsewhere in the notebook that already contains the answer. A function that does so would score well on "did the puzzle get solved" while the model learned nothing about solving Sudoku.

For this reason, model-written code is passed through a set of guards, provided by Unsloth, before it is ever executed:

- **`check_python_modules`** inspects the code and determines whether it imports anything outside an allowed set of built-ins. The `no_cheating` reward converts a violation into a penalty.
- **`create_locked_down_function`** compiles the function in a restricted namespace, preventing it from accessing the notebook's global variables.
- **`execute_with_time_limit`** bounds how long a strategy may run, so that an infinite loop costs a few seconds rather than stalling training indefinitely.

A second justification for the sandbox is independent of reward hacking: the notebook is about to execute code written by a language model, automatically, hundreds of times. Running such code without safeguards would be inadvisable regardless of the reward design.

<img alt="Without a sandbox the model can import an external solver and score highly without learning; with import blocking, a time limit and a locked-down function the only way to score is to solve the puzzle" src="" width="100%">

*Every loophole you leave open, the model will find.*

## LoRA and memory efficiency

Gemma 4 has billions of parameters. Updating all of them during training would require storing gradients and optimizer state for each one, exceeding the memory available on a single GPU.

**LoRA (Low-Rank Adaptation)** addresses this constraint. It freezes the original model weights and introduces small trainable "adapter" matrices into selected layers. Only the adapters are trained, typically 1 to 2 percent of the total parameters. This yields most of the quality of full fine-tuning at a small fraction of the memory cost, and produces a compact adapter file that can be shared and loaded on top of the base model.

The configuration used below sets `lora_rank = 32`, with adapters placed on the attention projections (`q_proj`, `k_proj`, `v_proj`, `o_proj`) and the MLP projections (`gate_proj`, `up_proj`, `down_proj`).

Unsloth supplies hand-optimized kernels for the training path, which substantially reduces memory use and increases training speed without loss of accuracy, including on AMD GPUs.

<img alt="Full fine-tuning updates every weight and needs large GPUs, while LoRA freezes the base model and trains small adapter matrices instead" src="" width="100%">

*This is why the whole workshop fits on a single GPU.*

# Part 2: The Sudoku environment

Before the model can be rewarded, an object for it to act on is required. In reinforcement learning terminology, this is the environment: the world the policy interacts with, and the component that determines whether an action was legal.

The environment implemented here is a Sudoku game with a deliberately small interface:

- `board()` returns the current 9x9 grid, where `0` denotes an empty cell.
- `initial_board()` returns the starting puzzle, allowing a strategy to determine which cells were given and must not be altered.
- `place_number(row, col, num)` attempts a move and returns whether it was legal.
- `state()` reports `ongoing`, `success`, or `failed`.

A move is legal only if the target cell is empty in both the current and the initial board, and the digit does not already appear in the corresponding row, column, or 3x3 box. A single illegal move ends the game, so a strategy cannot reach a solution by guessing.

````python
#@title Sudoku Game Implementation
from dataclasses import dataclass, field
from typing import List, Tuple, Optional
import random
import copy

def _is_valid_placement(board: List[List[int]], row: int, col: int, num: int) -> bool:
    """Check if placing num at (row, col) is valid."""
    # Check row
    if num in board[row]:
        return False

    # Check column
    if num in [board[r][col] for r in range(9)]:
        return False

    # Check 3x3 box
    box_row, box_col = 3 * (row // 3), 3 * (col // 3)
    for r in range(box_row, box_row + 3):
        for c in range(box_col, box_col + 3):
            if board[r][c] == num:
                return False

    return True

def _solve_sudoku(board: List[List[int]]) -> bool:
    """Solve sudoku using backtracking (for puzzle generation)."""
    for row in range(9):
        for col in range(9):
            if board[row][col] == 0:
                for num in range(1, 10):
                    if _is_valid_placement(board, row, col, num):
                        board[row][col] = num
                        if _solve_sudoku(board):
                            return True
                        board[row][col] = 0
                return False
    return True

def _generate_complete_board(rng: random.Random) -> List[List[int]]:
    """Generate a complete valid Sudoku board."""
    board = [[0 for _ in range(9)] for _ in range(9)]

    # Fill diagonal 3x3 boxes first (they don't affect each other)
    for box in range(3):
        nums = list(range(1, 10))
        rng.shuffle(nums)
        for i in range(3):
            for j in range(3):
                board[box * 3 + i][box * 3 + j] = nums[i * 3 + j]

    # Solve the rest
    _solve_sudoku(board)
    return board

@dataclass
class SudokuGame:
    difficulty: int = 40  # Number of cells to remove (20=easy, 40=medium, 50=hard)
    seed: Optional[int] = None
    _rng: random.Random = field(init=False, repr=False)
    _board: List[List[int]] = field(init=False, repr=False)
    _solution: List[List[int]] = field(init=False, repr=False)
    _initial_board: List[List[int]] = field(init=False, repr=False)
    _moves: int = field(default=0, init=False, repr=False)
    _state: str = field(default="ongoing", init=False, repr=False)

    def __post_init__(self):
        self._rng = random.Random(self.seed)

        # Generate complete board
        complete_board = _generate_complete_board(self._rng)
        self._solution = copy.deepcopy(complete_board)

        # Remove cells to create puzzle
        self._board = copy.deepcopy(complete_board)
        cells = [(r, c) for r in range(9) for c in range(9)]
        self._rng.shuffle(cells)

        for r, c in cells[:self.difficulty]:
            self._board[r][c] = 0

        self._initial_board = copy.deepcopy(self._board)
        self._update_state()

    def board(self) -> List[List[int]]:
        """Return current board state."""
        return [row[:] for row in self._board]

    def initial_board(self) -> List[List[int]]:
        """Return initial puzzle state."""
        return [row[:] for row in self._initial_board]

    def state(self) -> str:
        """Return game state: 'ongoing', 'success', or 'failed'."""
        return self._state

    def moves(self) -> int:
        """Return number of moves made."""
        return self._moves

    def place_number(self, row: int, col: int, num: int) -> bool:
        """Place a number on the board. Returns True if valid move."""
        # Validate input
        if not (0 <= row < 9 and 0 <= col < 9):
            self._state = "failed"
            return False

        if not (1 <= num <= 9):
            self._state = "failed"
            return False

        # Can't modify initial cells
        if self._initial_board[row][col] != 0:
            self._state = "failed"
            return False
        if self._board[row][col] != 0:
            self._state = "failed"
            return False
        # Check if placement is valid
        if not _is_valid_placement(self._board, row, col, num):
            self._state = "failed"
            return False

        # Place number
        self._board[row][col] = num
        self._moves += 1
        self._update_state()
        return True

    def _update_state(self) -> None:
        """Update game state based on current board."""
        # Check if puzzle is complete
        if all(self._board[r][c] != 0 for r in range(9) for c in range(9)):
            # Verify solution is correct
            if self._board == self._solution:
                self._state = "success"
            else:
                self._state = "failed"
        else:
            self._state = "ongoing"

    def pretty(self, colors: bool = True) -> str:
        """Pretty print the Sudoku board."""
        RESET = "\x1b[0m"
        INITIAL = "\x1b[38;5;45m"   # Cyan for initial numbers
        PLACED = "\x1b[38;5;196m"    # Red for placed numbers
        EMPTY = "\x1b[38;5;239m"     # Gray for empty cells

        lines = []
        lines.append("┌───────┬───────┬───────┐")

        for row in range(9):
            row_str = "│ "
            for col in range(9):
                num = self._board[row][col]

                if colors:
                    if num == 0:
                        row_str += f"{EMPTY}.{RESET}"
                    elif self._initial_board[row][col] != 0:
                        row_str += f"{INITIAL}{num}{RESET}"
                    else:
                        row_str += f"{PLACED}{num}{RESET}"
                else:
                    row_str += str(num) if num != 0 else "."

                if col % 3 == 2:
                    row_str += " │ "
                else:
                    row_str += " "

            lines.append(row_str.rstrip())

            if row == 8:
                lines.append("└───────┴───────┴───────┘")
            elif row % 3 == 2:
                lines.append("├───────┼───────┼───────┤")

        return "\n".join(lines)
````

# Part 3: Executing model-written code safely

This section constructs the harness that takes a strategy function and plays it against a game, converting "the model produced some code" into a numerical score.

`_execute_strategy` proceeds in a loop: it requests a move from the strategy, validates the shape of the returned value, attempts the move, and repeats. The loop terminates on the first illegal move or malformed return value, and is capped at 100 moves so that a strategy which repeats the same move indefinitely cannot stall execution.

The function returns both the number of valid moves made and the final game state. Partial credit is important in reinforcement learning: a strategy that makes 12 legal moves before failing is meaningfully better than one that fails immediately, and the reward should reflect this distinction.

````python
from typing import Callable
from unsloth import execute_with_time_limit

def _execute_strategy(strategy: Callable, game: SudokuGame):
    """Execute a strategy function on a Sudoku game."""
    assert callable(strategy)

    max_moves = 100
    valid_moves = 0  # Track successful moves

    while game.state() == "ongoing" and valid_moves < max_moves:
        try:
            board = game.board()
            initial = game.initial_board()
            result = strategy(board, initial)

            # Validate result format
            if not isinstance(result, (tuple, list)) or len(result) != 3:
                # Invalid format = immediate fail, but return valid moves made
                return valid_moves, "failed"

            row, col, num = result

            # Validate types
            if not all(isinstance(x, int) for x in [row, col, num]):
                return valid_moves, "failed"

            # Try to place number
            success = game.place_number(row, col, num)

            if success:
                valid_moves += 1  # Count this valid move
            else:
                # Invalid move = game fails, but return valid_moves made so far
                return valid_moves, "failed"

        except Exception:
            return valid_moves, "failed"

    if valid_moves >= max_moves and game.state() == "ongoing":
        return valid_moves, "failed"

    return valid_moves, game.state()
````

This loop is wrapped in a time limit next. Model-generated code may contain infinite loops, and without a bound a single defective strategy would stall training.

The original workshop specification uses a 10-second limit to permit more elaborate strategies. A 5-second limit is used here to keep the session brief; this value may be increased to allow the model more time for slower, more thorough search.

````python
@execute_with_time_limit(5)  # Set a time limit of 5 seconds for strategy execution, originally was 10 seconds
def execute_strategy(strategy: Callable, game: SudokuGame):
    """Execute strategy with 10 second time limit."""
    return _execute_strategy(strategy, game)
````

The harness is checked against a trivial hand-written strategy first, to confirm that the machinery functions correctly before it is given model-generated output.

````python
def simple_strategy(board, initial):
    """Simple strategy: fill first empty cell with 1."""
    for r in range(9):
        for c in range(9):
            if board[r][c] == 0 and initial[r][c] == 0:
                return (r, c, 7)
    return (0, 0, 7)

game = SudokuGame(difficulty=30, seed=42)
try:
    moves, state = execute_strategy(simple_strategy, game)
    print(f"Moves: {moves}, State: {state}")
except TimeoutError as e:
    print(f"Timed out: {e}")
````

# Part 4: The prompt and the untrained baseline

The prompt constitutes the task specification. It communicates to the model the shape of the input, the shape of the expected output, the rules that must be respected, and the required answer format (a Python function enclosed in triple backticks, on which the extraction step depends).

This prompt is the natural starting point for adapting the notebook to a different reinforcement learning task: the task description, together with the corresponding reward functions, can be changed while the remainder of the machinery carries over unchanged.

````python
prompt = """
Create a Sudoku solving strategy using only native Python built-in functions without any import statements.
You are given two lists of lists (9x9 grids):
- board: current state (0 means empty)
- initial: starting puzzle (0 means was empty, numbers are fixed)

Return a tuple (row, col, number) for the next move.
- row: 0-8 (row index)
- col: 0-8 (column index)
- number: 1-9 (digit to place)

Only place numbers in cells that are BOTH empty in initial AND empty in board (initial[row][col] == 0 AND board[row][col] == 0)
Use Sudoku rules: no duplicates in rows, columns, or 3x3 boxes.
Output your function in backticks:
```python
def strategy(board, initial):
    # Your logic here
    return (row, col, number)
```
All helper functions must be inside def strategy. Output only the function.
""".strip()

print(prompt)
````

Before training, the model's unaided output is examined. This cell should be run and its output read carefully, since it establishes the baseline: the same prompt is run again after training, for comparison.

Typical failure modes at this stage include prose in place of code, a function that does not respect the `initial` board, or a plausible-looking solver that makes an illegal move on its first attempt.

# Part 5: The reward functions

Every reward function follows the contract required by the GRPO trainer: it takes a list of `completions` and returns a list of scores, one per completion. The trainer sums the scores returned by all three functions.

A helper function, `extract_function`, is defined first; it extracts the Python code from the triple backticks the model was instructed to use.

````python
def extract_function(text):
    """Extract Python function from markdown code blocks."""
    if text.count("```") >= 2:
        first = text.find("```") + 3
        second = text.find("```", first)
        fx = text[first:second].strip()
        fx = fx.removeprefix("python\n")
        fx = fx[fx.find("def"):]
        if fx.startswith("def strategy(board, initial):"):
            return fx
    return None
````

### Reward 1: does the code execute?

`function_works` addresses the most basic requirement: whether the completion contains valid, runnable Python code defining a `strategy` function.

- `+1.0` if the code compiles into a locked-down function.
- `-1.0` if a function was found but fails to compile.
- `-2.0` if no usable function could be extracted at all.

Without this reward term, the model has no incentive to produce parseable code, and the remaining two rewards would have nothing to evaluate.

````python
def function_works(completions, **kwargs):
    """Reward for generating valid executable Python code."""
    scores = []
    for completion in completions:
        score = 0
        response = completion[0]["content"]
        function = extract_function(response)

        if function is not None:
            ok, info = check_python_modules(function)

        if function is None or "error" in info:
            score = -2.0  # Invalid function
        else:
            try:
                new_strategy = create_locked_down_function(function)
                score = 1.0  # Valid function
            except:
                score = -1.0  # Function has errors

        scores.append(score)
    return scores
````

### Reward 2: no cheating

`no_cheating` uses `check_python_modules` to detect imports and applies a substantial penalty for their presence: `-20.0`, compared with `+1.0` for code containing none.

The magnitude of this penalty is a deliberate design choice. Solving the puzzle is worth `+30.0` under Reward 3. If cheating incurred only a small penalty, importing a solver would remain the highest-scoring action available, and the model would learn to cheat. The penalty must therefore be large enough that the shortcut is never advantageous: the disallowed path must not out-score the intended one.

````python
def no_cheating(completions, **kwargs):
    """Penalize use of external imports."""
    scores = []
    for completion in completions:
        response = completion[0]["content"]
        function = extract_function(response)

        if function is not None:
            ok, info = check_python_modules(function)
            scores.append(1.0 if ok else -20.0)  # Heavy penalty for cheating
        else:
            scores.append(-1.0)  # Failed to create function

    return scores
````

### Reward 3: does the strategy solve the puzzle?

This reward corresponds to the primary objective. `strategy_succeeds` extracts the function, re-checks it for imports, locks it down, constructs a fresh puzzle, and executes it.

The scoring is designed to provide a gradient rather than a single all-or-nothing outcome:

| Outcome | Score |
| --- | --- |
| Puzzle solved | `+30.0` |
| Some legal moves before failing | `+0.2` per valid move |
| No legal moves at all | `-2.0` |
| Timed out | `-1.0` |
| Raised an exception | `-3.0` |

This partial credit is what makes learning tractable. If only a complete solution were scored, nearly every early attempt would receive a score of zero, the group would exhibit no variance, and GRPO would have no basis for comparison. By awarding credit per legal move, a strategy that reaches 5 moves is measurably better than one that reaches 2, giving the model a signal to improve on.

All completions within a batch are scored on the same puzzle, since `seed` is drawn once per call. This ensures that the comparison within a group is fair, which is a requirement for GRPO.

The cell additionally prints a sample of generated strategies as training proceeds, allowing the evolution of the model's code to be observed directly.

````python
import numpy as np

global PRINTER
PRINTER = 0

def strategy_succeeds(completions, **kwargs):
    """Reward valid moves even if strategy eventually fails."""
    global PRINTER
    scores = []

    seed = np.random.randint(10000)
    difficulty = 40
    for completion in completions:
        printed = False
        response = completion[0]["content"]
        function = extract_function(response)

        if PRINTER % 5 == 0:
            printed = True
            print("\n" + "=" * 60)
            print(function)
            print("=" * 60)
        PRINTER += 1

        if function is not None:
            ok, info = check_python_modules(function)

        if function is None or "error" in info:
            scores.append(0)
            continue

        try:
            new_strategy = create_locked_down_function(function)
        except:
            scores.append(0)
            continue

        try:
            game = SudokuGame(difficulty=difficulty, seed=seed)
            valid_moves, game_state = execute_strategy(new_strategy, game)
            if valid_moves == difficulty:
                game_state = "success"

            print(f"\n Valid moves: {valid_moves}, Final state: {game_state}")

            if not printed:
                print("Strategy:")
                print(function[:200] + "..." if len(function) > 200 else function)

            print("\nFinal board:")
            print(game.pretty())

            if game_state == "success":
                scores.append(30.0)  # Solved the puzzle!
            elif valid_moves > 0:
                # Reward based on valid moves made before failure
                # Each valid move is worth 0.2 points
                reward = valid_moves * 0.2
                scores.append(reward)
            else:
                scores.append(-2.0)  # Failed immediately with no valid moves

        except TimeoutError:
            print("Timeout")
            scores.append(-1.0)
        except Exception as e:
            print(f"Exception: {str(e)[:100]}")
            scores.append(-3.0)

    return scores
````

# Part 6: Training

## The dataset

The dataset used for GRPO training consists solely of prompts. It contains no target answers, since nothing in this pipeline compares the model's output to a reference solution. All learning signal is derived from executing the generated code and scoring the result.

````python
from datasets import Dataset

dataset = Dataset.from_list([
    {
        "prompt": [{"role": "user", "content": prompt.strip()}],
        "answer": 0,
    }
] * 1000)

maximum_length = len(tokenizer.apply_chat_template(
    [{"role": "user", "content": prompt.strip()}],
    add_generation_prompt=True
))

print(f"Maximum prompt length: {maximum_length}")
print("\nDataset sample:")
print(dataset[0])
````

## Trainer configuration

Several settings merit explanation, as they are the parameters most likely to be adjusted in subsequent experiments:

- **`num_generations = 2`** sets the group size: the number of candidates generated per prompt and compared against one another. This corresponds to the "G" in GRPO. A value of 2 is the minimum practical for this workshop; larger groups yield a stronger comparison signal and typically improve learning, at higher memory cost.
- **`max_steps = 15`** keeps the demonstration brief. The original specification uses 60 steps, and full training runs use substantially more.
- **`learning_rate = 1e-4`** is increased from the original value of `5e-5` to produce visible progress within a short run.
- **`temperature = 1.0`** maintains diversity in generation. This matters more in reinforcement learning than in standard inference: if every candidate within a group were identical, there would be no basis for comparison and no learning signal.
- **`mask_truncated_completions = True`** excludes completions that were cut off by the length limit, so that the model is not penalized for an incomplete answer it was never given the opportunity to finish.

````python
# Leave room for the prompt (plus 1 token safety margin)
max_completion_length = max_seq_length - (maximum_length + 1)

from trl import GRPOConfig, GRPOTrainer
training_args = GRPOConfig(
    temperature = 1.0,
    learning_rate = 1e-4, #originally was 5e-5
    weight_decay = 0.001,
    warmup_ratio = 0.1,
    lr_scheduler_type = "linear",
    optim = "adamw_8bit",
    logging_steps = 1,
    per_device_train_batch_size = 1,
    gradient_accumulation_steps = 2, # Increase to 4 for smoother training
    num_generations = 2, # Decrease if out of memory
    max_completion_length = max_completion_length,
    # num_train_epochs = 1, # Set to 1 for a full training run
    max_steps = 15, #originally 60,
    save_steps = 100,
    report_to = "none", # Can use Weights & Biases, TrackIO
    output_dir = "outputs",
    epsilon=0.2,
    epsilon_high=0.28, # one sided
    delta=1.5, # two sided
    loss_type='bnpo',
    mask_truncated_completions=True
    # For optional training + evaluation
    # fp16_full_eval = True,
    # per_device_eval_batch_size = 4,
    # eval_accumulation_steps = 1,
    # eval_strategy = "steps",
    # eval_steps = 1,
)
````

The trainer is assembled next, given the three reward functions as a list. The trainer calls each function on every completion and sums the resulting scores.

````python
# For optional training + evaluation
# new_dataset = dataset.train_test_split(test_size = 0.01)

trainer = GRPOTrainer(
    model = model,
    processing_class = tokenizer,
    reward_funcs = [
        function_works,
        no_cheating,
        strategy_succeeds,
    ],
    args = training_args,
    train_dataset = dataset,

    # For optional training + evaluation
    # train_dataset = new_dataset["train"],
    # eval_dataset = new_dataset["test"],
)
````

## Running the training

Training is executed in this cell. The `reward` column in the log is the value to monitor; it is expected to trend upward over time.

The following behavior should be expected, and should not be mistaken for a malfunction:

- Early rewards are frequently negative, as the model initially produces unusable code or code that imports disallowed modules, incurring penalties.
- Progress is not smooth: the reward varies between steps because each step evaluates a different puzzle and a small group of samples.
- With `max_steps = 15`, the training run constitutes a demonstration rather than a converged model. Full training runs typically require 150 to 200 steps before the upward trend becomes unambiguous.

The printed strategies are also informative: the qualitative shift from prose or broken code toward a function that correctly checks rows, columns, and boxes is often more informative than the numerical reward.

The first few rows of output resemble the following: the loss remains at zero, and the reward is small and noisy. This is expected behavior.

| Step | Training Loss | reward    | reward_std | completion_length | kl       |
|------|---------------|-----------|------------|-------------------|----------|
| 1    | 0.000000      | 0.125000  | 0.000000   | 200.000000        | 0.000000 |
| 2    | 0.000000      | 0.072375  | 0.248112   | 200.000000        | 0.000000 |
| 3    | 0.000000      | -0.079000 | 0.163776   | 182.500000        | 0.000005 |

In GRPO, the loss value does not serve as a measure of quality in the way it does in supervised training. The reward column should be used to assess progress instead.

````python
trainer.train()
````

## Resuming training from a checkpoint

Training runs are not always completed in a single sitting: a session may end, a machine may be reassigned, or a longer run may need to proceed in stages. `GRPOTrainer` saves a checkpoint automatically when training completes, and that checkpoint can be used to continue training later without starting over.

The cell below builds a new `GRPOConfig` with `max_steps` raised from 15 to 20, then resumes from the checkpoint the previous cell already saved (`outputs/checkpoint-15`). Two things confirm the resume actually worked, rather than silently restarting from scratch: the progress bar should begin at step 16 rather than step 1, and the learning rate should continue decaying from where the original schedule left off rather than restarting its warmup.

````python
resume_training_args = GRPOConfig(
    temperature = 1.0,
    learning_rate = 1e-4,
    weight_decay = 0.001,
    warmup_ratio = 0.1,
    lr_scheduler_type = "linear",
    optim = "adamw_8bit",
    logging_steps = 1,
    per_device_train_batch_size = 1,
    gradient_accumulation_steps = 2,
    num_generations = 2,
    max_completion_length = max_completion_length,
    max_steps = 20,  # 5 additional steps on top of the checkpoint already saved at step 15
    save_steps = 100,
    report_to = "none",
    output_dir = "outputs",
    epsilon = 0.2,
    epsilon_high = 0.28,
    delta = 1.5,
    loss_type = 'bnpo',
    mask_truncated_completions = True,
)

resume_trainer = GRPOTrainer(
    model = model,
    processing_class = tokenizer,
    reward_funcs = [
        function_works,
        no_cheating,
        strategy_succeeds,
    ],
    args = resume_training_args,
    train_dataset = dataset,
)

resume_trainer.train(resume_from_checkpoint = "outputs/checkpoint-15")
print("Reached global step:", resume_trainer.state.global_step)
````

## Checking that the model is still a generalist

Only 1.16% of the model's parameters were updated, for a small number of steps, using a reward signal narrowly focused on one task: writing a working Sudoku strategy function. It is worth checking directly that this specialization has not come at the cost of the model's general abilities, and that training did not overwrite its capacity for ordinary conversation or unrelated coding tasks.

The three prompts below have nothing to do with Sudoku. The second reuses the "no imports" constraint from the training prompt on a different coding task; the third removes that constraint entirely, on a task (maze-solving) where an import is the natural choice. If the model answers all three sensibly, and only avoids imports when the prompt actually asks for it, that is evidence the reward signal shaped a narrow skill without corrupting the model's broader competence.

````python
def ask(question, max_new_tokens = 200):
    chat_text = tokenizer.apply_chat_template(
        [{"role": "user", "content": question}],
        tokenize = False,
        add_generation_prompt = True,
    )
    inputs = tokenizer(images=None, text=chat_text, return_tensors = "pt").to("cuda")
    output = model.generate(
        **inputs,
        max_new_tokens = max_new_tokens,
        temperature = 0.7,
        do_sample = True,
    )
    return tokenizer.decode(output[0][inputs["input_ids"].shape[1]:], skip_special_tokens = True)

print("--- General knowledge ---")
print(ask("What is the capital of France, and what is one famous landmark there?"))

print("\n--- A different coding task, same 'no imports' constraint ---")
print(ask("Write a Python function that reverses a string using only built-in functions, no imports."))

print("\n--- A different puzzle-solving task, no import restriction given ---")
print(ask("Write a Python function that solves a simple maze represented as a 2D grid, where 0 is open and 1 is a wall. Return the path as a list of (row, col) tuples."))
````
